# Chapter 00：环境检查

**目标**：确认 PyTorch、CUDA、Triton 和 Jupyter 环境可用。本章不写 Triton kernel。

运行前需要 Python 3.10+、CUDA 版 PyTorch、Triton 和 NVIDIA GPU。

## 1. 导入并查看版本

如果这里导入失败，先回到项目根目录安装 `requirements.txt`。

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name.startswith("chapter_"):
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import torch
import triton

print(f"PyTorch version: {torch.__version__}")
print(f"Triton version: {triton.__version__}")
print(f"CUDA runtime: {torch.version.cuda}")
```

## 2. 检查 CUDA 和 GPU

`torch.cuda.is_available()` 必须为 `True`。否则通常是没有 NVIDIA GPU，或安装了 CPU 版 PyTorch。

```python
if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA is not available. Install CUDA-enabled PyTorch and use an NVIDIA GPU."
    )

device = torch.device("cuda")
print(f"CUDA available: {torch.cuda.is_available()}")
print(f"GPU: {torch.cuda.get_device_name(torch.cuda.current_device())}")
```

## 3. 创建 CUDA tensor 并运行 PyTorch 运算

tensor 的 `device` 应显示为 `cuda:0`，加法结果应为 `[5, 7, 9]`。

```python
x = torch.tensor([1.0, 2.0, 3.0], device=device)
y = torch.tensor([4.0, 5.0, 6.0], device=device)
z = x + y
print(f"x.device = {x.device}")
print(f"z = {z}")
assert torch.equal(z, torch.tensor([5.0, 7.0, 9.0], device=device))
print("Triton tutorial environment is ready")
```
